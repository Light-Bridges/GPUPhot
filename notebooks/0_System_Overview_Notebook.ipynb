# System Overview
This notebook provides an overview of the system architecture and services available in the GPUPhot Docker Compose setup.

## Services Overview
The system is composed of the following services:

1. **RabbitMQ**: Message broker for Celery task queue.
   - Ports: `5672` (AMQP) and `15672` (Management UI).
2. **Redis**: Backend for Celery task results.
   - Port: `6379`.
3. **Celery Beat**: Scheduler for periodic tasks.
4. **GPUPhot Worker**: Celery worker for processing astronomical images.
   - Supports GPU acceleration.
5. **JupyterLab**: Interactive environment for running notebooks.
   - Port: `8888`.
6. **Flower**: Monitoring tool for Celery tasks.
   - Port: `5555`.
7. **PostgreSQL**: Database for storing processing results.
   - Port: `5432`.
